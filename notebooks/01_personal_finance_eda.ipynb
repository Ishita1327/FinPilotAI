```python

import pandas as pd
from pathlib import Path

# Identify the project directory
project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

# Locate the Kaggle dataset
file_path = (
    project_root
    / "data"
    / "raw"
    / "personal_finance"
    / "Personal_Finance_Dataset.csv"
)

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

display(df.head())
```

```python

# Column names
print("COLUMN NAMES")
print(df.columns.tolist())

# Data types
print("\nDATA TYPES")
print(df.dtypes)

# Missing values
print("\nMISSING VALUES")
print(df.isnull().sum())

# Duplicate rows
print("\nDUPLICATES")
print(df.duplicated().sum())

# Transaction types
print("\nTRANSACTION TYPES")
print(df["Type"].value_counts())

# Categories
print("\nCATEGORIES")
print(df["Category"].value_counts())
```

```python

print("AMOUNT SUMMARY")
print(df["Amount"].describe())

print("\nDATE RANGE")
dates = pd.to_datetime(df["Date"], errors="coerce")

print("Earliest date:", dates.min())
print("Latest date:", dates.max())
print("Invalid dates:", dates.isna().sum())

print("\nAMOUNT ISSUES")
amounts = pd.to_numeric(df["Amount"], errors="coerce")

print("Invalid amounts:", amounts.isna().sum())
print("Negative amounts:", (amounts < 0).sum())
print("Zero amounts:", (amounts == 0).sum())
```

```python

summary = pd.DataFrame({
    "column": df.columns,
    "data_type": df.dtypes.astype(str).values,
    "missing_values": df.isnull().sum().values,
    "missing_percent": (
        df.isnull().mean() * 100
    ).round(2).values,
    "unique_values": df.nunique().values
})

display(summary)
```

```python

# Load the second personal transactions dataset

excel_path = (
    project_root
    / "data/raw/personal_finance/personal_transactions.xlsx"
)

df_excel = pd.read_excel(excel_path)

print("Dataset shape:", df_excel.shape)

print("\nDate range:")
print("Start:", df_excel["Date"].min())
print("End:", df_excel["Date"].max())

print("\nTransaction types:")
print(df_excel["Transaction Type"].value_counts())

print("\nTop 20 transaction descriptions:")
print(df_excel["Description"].value_counts().head(20))

print("\nAccounts:")
print(df_excel["Account Name"].value_counts())

print("\nCategories:")
print(df_excel["Category"].value_counts())
```

```python

# Inspect credit card payments
card_payments = df_excel[
    df_excel["Description"]
    .str.contains("Credit Card Payment", case=False, na=False)
]

print("Credit card payment records:", len(card_payments))

display(
    card_payments[
        ["Date", "Description", "Amount",
         "Transaction Type", "Category", "Account Name"]
    ].head(15)
)

# Inspect potential subscriptions
subscriptions = df_excel[
    df_excel["Description"]
    .str.contains(
        "Spotify|Netflix|Subscription",
        case=False,
        na=False
    )
]

print("\nPotential subscription records:", len(subscriptions))

display(
    subscriptions[
        ["Date", "Description", "Amount",
         "Transaction Type", "Category"]
    ].sort_values("Date").head(30)
)
```

```python

# Merchant recurrence summary
merchant_summary = (
    df_excel
    .groupby("Description")
    .agg(
        transaction_count=("Amount", "count"),
        average_amount=("Amount", "mean"),
        minimum_amount=("Amount", "min"),
        maximum_amount=("Amount", "max"),
        first_date=("Date", "min"),
        last_date=("Date", "max")
    )
    .sort_values("transaction_count", ascending=False)
)

display(merchant_summary.head(20))
```
